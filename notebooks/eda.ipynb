# Exploratory Data Analysis of LLM Inference Trace Data

```python
import pandas as pd
import numpy as np
import boto3
import io
```

```python
# pulling raw data from AWS S3 bucket
s3 = boto3.client('s3', region_name='us-east-2')

response = s3.get_object(Bucket='amzn-s3-azurellm-bucket', Key='AzureLLMInferenceTrace_code_1week.csv')
df = pd.read_csv(io.BytesIO(response['Body'].read()))

print(df.head())

```
